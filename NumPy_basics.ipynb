```python
# Activity 2 : perform basic matrix operation
import numpy as np
a = np.array([[1,2] , [2,3]])
b = np.array([[2,3] , [6,7]])
print("matrix a: ")
print(a)

print("matrix b: ")
print(b)

print("\naddition: ")
print(a+b)

print("\nsubtraction: ")
print(a-b)

print("\nelement wise multiplication: ")
print(a*b)

print("\nmatrix wise multiplication: ")
print(a@b)

```

```python
"""weather data of july days in udupi
find hottest and coldest day
avg temp
dates above 35 deg cel"""

# July temperatures (Max, Min)
temp = [
    (26,23), (28,24), (28,24), (29,24), (30,25),
    (30,24), (30,23), (30,24), (31,26), (30,25),
    (30,24), (30,24), (30,26), (31,25), (31,25),
    (30,24), (27,22), (29,23), (29,25), (27,24),
    (29,24), (26,24), (29,23), (30,24), (29,23),
    (29,24), (30,24), (28,24), (29,25), (30,24),
    (30,25)
]

# Average temperature for each day
avg = []
for t in temp:
    avg.append((t[0] + t[1]) / 2)

print("Daily Average Temperatures:")
print(avg)

# Monthly average
monthly_avg = sum(avg) / len(avg)
print("Monthly Average:", round(monthly_avg, 2), "°C")

# Hottest day
max_temp = max(t[0] for t in temp)
print("Highest Temperature:", max_temp, "°C")

print("Hottest Day(s):")
for i in range(len(temp)):
    if temp[i][0] == max_temp:
        print("July", i + 1)

# Coldest day
min_temp = min(t[0] for t in temp)
print("Lowest Temperature:", min_temp, "°C")

print("Coldest Day(s):")
for i in range(len(temp)):
    if temp[i][0] == min_temp:
        print("July", i + 1)

# Days above 35°C
print("Days Above 35°C:")
found = False
for i in range(len(temp)):
    if temp[i][0] > 35:
        print("July", i + 1)
        found = True

if not found:
    print("None")

```
