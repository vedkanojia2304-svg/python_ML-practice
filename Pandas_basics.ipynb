```python
""" Assignment
use titanic dataset from kaggle
find the total no. of passengers
find the avg age of surviors and non survivors
list the passenger which are older than 60 years
find the avg fare of each passenger """

# Reading the csv file
import kagglehub
import pandas as pd
import os

# Download latest version
path = kagglehub.dataset_download("yasserh/titanic-dataset")

print("Path to dataset files:", path)

# Show the files inside the folder
print(os.listdir(path))
```

```python
# loading the file

df = pd.read_csv(os.path.join(path, "Titanic-Dataset.csv"))

# Display first 5 rows
df.head()
```

```python
# find the total no. of passengers

print("Total no. of passengers" , len(df))
```

```python
# Average age of survivors
print("\nAverage age of survivors:")
print(df[df["Survived"] == 1]["Age"].mean())
```

```python
# Average age of non survivors

print("\nAverage age of non survivors:")
print(df[df["Survived"] == 0]["Age"].mean())
```

```python
# avg fare of each passenger

print(df.groupby("Survived")["Fare"].mean())

# print("\nAverage fare for each passenger class:")
# print(df.groupby("Pclass")["Fare"].mean())
```
